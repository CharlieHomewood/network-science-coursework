```python
from matplotlib.offsetbox import AnchoredText
import matplotlib.pyplot as plt
import networkx as nx
import polars as pl
import seaborn as sns
```

```python
df = pl.read_parquet("../data/events_data.parquet")

pass_df = df.filter(pl.col("type") == "Pass")

pass_df = pass_df.select([col for col in pass_df.columns if not pass_df[col].is_null().all()])
```

```python
average_number_of_events_df = df.group_by("match_id", maintain_order=True).len()

sns.histplot(
    data=average_number_of_events_df,
    x="len"
)

mean = average_number_of_events_df["len"].mean()
sd = average_number_of_events_df["len"].std()

textstr = f"$\\bar{{x}}$ = {mean:.2f}\n$\\sigma$ = {sd:.2f}"

at = AnchoredText(textstr, loc="upper left", prop=dict(size=12), frameon=True)
plt.gca().add_artist(at)

plt.title("Average Number of Events per Match", weight="bold")
plt.xlabel("Number of Events", weight="bold")
plt.ylabel("Frequency", weight="bold")

plt.show()
```
